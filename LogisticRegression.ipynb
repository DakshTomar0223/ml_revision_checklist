```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import fetch_openml
```

```python
titanic_data = sns.load_dataset('titanic')
```

```python
df = titanic_data.drop('survived', axis=1).copy()
```

```python
y = titanic_data['survived'].copy()
```

```python
df
```

```python
df['family'] = df['sibsp'] + df['parch']
df.loc[df['family'] > 0, 'travelled_alone'] = 0
df.loc[df['family']==0, 'travelled_alone'] = 1
df['family'].head()
```

```python
df.drop(['sibsp','parch'],axis=1, inplace = True)
```

```python
df
```

```python
from sklearn.impute import SimpleImputer
imp_mean = SimpleImputer(strategy = 'mean')
```

```python
df['age'] = imp_mean.fit_transform(df[['age']])
df['fare'] = imp_mean.fit_transform(df[['fare']])
```

```python
imp_freq = SimpleImputer(strategy = 'most_frequent')
```

```python
df['embarked']= imp_freq.fit_transform(df[['embarked']]).flatten()
```

```python
embark = pd.get_dummies(df['embarked'], drop_first=True)
embark
```

```python
df = pd.concat([df, embark], axis=1)
df.drop(['embarked'], axis=1, inplace=True)
```

```python
df
```

```python
df.drop(['embark_town'], axis=1, inplace=True)
display(df.head())
```

```python
df['sex'] = df['sex'].map({'male': 0, 'female': 1})
df['adult_male'] = df['adult_male'].astype(int)
df['alone'] = df['alone'].astype(int)
df['alive'] = df['alive'].map({'no': 0, 'yes': 1})
display(df.head())
```

```python
df['class'] = df['class'].map({'First': 1, 'Second': 2, 'Third': 3})
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
df['who'] = le.fit_transform(df['who'])
display(df.head())
```

```python
df['deck'] = df['deck'].cat.add_categories('U').fillna('U')
df['deck'] = le.fit_transform(df['deck'])
display(df.head())
```

```python
print(df.info())
```

```python
df['Q'] = df['Q'].astype(int)
df['S'] = df['S'].astype(int)

print(df.info())
display(df.head())
```

```python
X = df.drop(['alive'],axis = 1)
```

```python
y = df['alive']
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.3,random_state=1)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
mod = LogisticRegression()
mod.fit(X_train,y_train)
```

```python
pred = mod.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test,pred)
```

```python
from sklearn.metrics import confusion_matrix
confusion_matrix(y_test,pred)
```
