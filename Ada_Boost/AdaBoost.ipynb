```python
import pandas as pd
```

```python
df = pd.read_csv('Titanic_test.csv')
```

```python
df
```

```python
df1 = pd.read_csv('Titanic_train.csv')
```

```python
df1
```

```python
datas = pd.concat([df,df1])
```

```python
datas
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
datas.isnull().sum()
```

```python
datas = datas.drop('Cabin',axis=1)
```

```python
from sklearn.impute import SimpleImputer
si = SimpleImputer(strategy='mean')
datas['Age'] = si.fit_transform(datas[['Age']])
datas = datas.drop('Embarked',axis=1)
```

```python
datas.isnull().sum()
```

```python
datas = datas.drop(['Name','Ticket'],axis=1)
datas
```

```python
threshold = 3
from scipy.stats import zscore
numerical_features = ['Fare','Age']
z_scores = np.abs(zscore(datas[numerical_features]))
print(z_scores)
```

```python
outliers = np.where(z_scores > threshold)
print(outliers)
```

```python
from scipy.stats.mstats import winsorize
datas['Age'] = winsorize(datas['Age'],limits = (0.15,0.15))
datas['Fare'] = winsorize(datas['Fare'],limits = (0.15,0.15))
```

```python
object_cols = datas.select_dtypes(include='object').columns
display(object_cols)
```

```python
datas = pd.get_dummies(datas, columns=object_cols, drop_first=True)
display(datas.head())
```

```python
datas['Sex_male'] = datas['Sex_male'].astype(int)
display(datas.head())
```

```python
X = datas.drop(['Survived'],axis = 1)
y = datas['Survived']
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.3,random_state=1)
```

```python
from sklearn.ensemble import AdaBoostClassifier
ab =AdaBoostClassifier(n_estimators=20,learning_rate=0.7,random_state=1)
ab.fit(X_train,y_train)
y_pred = ab.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)
```
